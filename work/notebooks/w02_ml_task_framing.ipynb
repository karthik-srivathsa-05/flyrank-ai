## 5. Why ML beats a fixed rule here

A fixed rule can capture an obvious pattern, such as prioritizing old pages with high impressions, but the relationship between several signals and declining performance is more complicated than one or two thresholds. ML can combine multiple observable signals and learn interactions that a simple if-statement may miss. The model still needs to be evaluated against a transparent baseline rather than assumed to be better.
